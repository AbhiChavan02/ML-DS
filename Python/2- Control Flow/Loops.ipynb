```python
## For Loop

for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)
```

```python
## For (String)

str = "Abhi Chavan"
for i in str:
    print(i)
```

```python
## While Loop

# The while loop continues to execute as long as the condition is True.

count = 0

while count<5:
    print(count)
    count=count+1
```

```python
count=0

while count%2==0:
    print(count)
    count=count+1
```

```python
## Loop Control Statements

#1. break - The break statement exits the loop prematurely
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
#2. continue - The continue statement skips the current iteration and continues with the next.

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
#3. pass - The pass statement is null operation; it does nothing.

for i in range(5):
    if i==3:
        print("The number is",i)
        pass
    print(i)
```

```python
## Nested Loop - a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples

#1. Calculate the sum of first N natural numbers using a while and for loop

n=10
sum=0
count=1

while count<=n:
    sum = sum+count
    count=count+1
print("The sum of first 10 natural number:",sum)
```

```python
n=10
sum=0
for i in range(11):
    sum=sum+i
print(sum)
```

```python
#2. Display the prime number between 1 and 100

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```
