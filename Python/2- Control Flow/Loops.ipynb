```python
## For Loop

for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)
```

```python
## For (String)

str = "Abhi Chavan"
for i in str:
    print(i)
```

```python
## While Loop

# The while loop continues to execute as long as the condition is True.

count = 0

while count<5:
    print(count)
    count=count+1
```

```python
count=0

while count%2==0:
    print(count)
    count=count+1
```

```python
## Loop Control Statements

#1. break - The break statement exits the loop prematurely
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
#2. continue - The continue statement skips the current iteration and continues with the next.

for i in range(10):
    if i%2==0:
        continue
    print(i)
```
