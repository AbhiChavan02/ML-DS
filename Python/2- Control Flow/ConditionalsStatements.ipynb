```python
## If Statement

age = 20
if age>=18:
    print("You are allowed to vote in the election")
```

```python
## Else Statement
# The else statement executes a block of code if the condition in the if statement is False.

age=16
if age>=18:
    print("You are allowed to vote in the election")
else:
    print("You are minor")
```

```python
## Elif Statement
# The elif statement allows you to check multiple conditions. It stands for "else if"

age = 20

if age<13:
    print("You are Child")
elif age<18:
    print("You are Teenager")
else:
    print("You are Adult")
```

```python
## Nested Conditional Statement
# You can place one or more if, elif, or else statements inside another if, elif, or else statement to create nested conditional statement

num = int(input("Enter the Number"))
if num>0:
    print("The number is positive")
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is zero or negative")
```

```python
## Practical Examples

#1. Determine if a year is a leap year using nested conditional statement
year = int(input("Enter the Year"))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"is a leap year")
        else:
            print(year,"is not a leap year")
    else:
        print(year,"is a leap year")
else:
    print(year,"is not a leap year")
```

```python
# Assignment
# Simple Calculator
num1 = float(input("Enter 1st Number"))
num2 = float(input("Enter 2nd Number"))
operation = input("Enter operation (+, -, *, /): ")

if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 != 0: 
        result = num1 / num2
    else:
        result = "Error! Divison by zero"
else:
    result = "Invalid Operation"

print("Num1:", num1)
print("Num2:", num2)
print("Operation:", operation)
print("Result:", result)
```
