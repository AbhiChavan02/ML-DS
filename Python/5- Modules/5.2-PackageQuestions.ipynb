```python
import math

print(math.sqrt(25))
print(math.sin(math.radians(90)))
```

6. Use the `os` module to create a new directory, list the contents of the current directory, and remove the newly created directory.


```python
import os

# Create a new dictionary
os.mkdir('new_dictionary')

# List contents of the current dictionary
print(os.listdir('.'))

# Remove the newly created dictionary
os.rmdir('new_dictionary')
print(os.listdir('.'))
```

8. Use the `math` module to calculate the greatest common divisor (GCD) of two numbers and the factorial of a number.


```python
import math

print(math.gcd(48, 18))
print(math.factorial(5))
```

9. Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.


```python
import datetime

# current date
today = datetime.date.today()
print(f"Today's date: {today}")

# Date 100 days to today
future_date = today + datetime.timedelta(days=100)
print(f"Date 100 days from today: {future_date}")

# Day of the week for a given date
given_date = datetime.date(2022, 2, 1)
print(f"Day of the week for 2022-02-01: {given_date.strftime('%A')}")
```

10. Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.


```python
import random

# List of 5 random numbers between 1 to 50
random_numbers = [random.randint(1, 50) for _ in range(5)]
print(random_numbers)

# Shuffle a list 
lst = [1, 2, 3, 4, 5]
random.shuffle(lst)
print(lst)
```
