#### Standard Library Overview


Python's Standard Library is a vast collection of modules and packages that come bundled with python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.


```python
import array

arr = array.array('i', [1,2,3,4])
print(arr)
```

```python
import math

print(math.sqrt(16))
print(math.pi)
```

```python
## random - randomly picking up any values
#1. randint - it means provide a any random integer based on the value that we give here
import random

print(random.randint(1,10))

#2. choice - it means if we have multiple choice here from that we have to select any choice
print(random.choice(['apple','banana','cherry']))
```

```python
### os - used for file and directory access

import os
print(os.getcwd())
```

```python
os.mkdir('test_dir')
```

```python
## High level operation on files and collection of files

import shutil
shutil.copyfile('source.txt','destination.txt')
```

```python
## Data Serialization - it is nothing data realization (json creation)
import json

data={'name':'Abhi','age':23}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))

past_data = json.loads(json_str)
print(past_data)
print(type(past_data))
```

```python
## CSV file 

import csv

with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Abhi', 24])

with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## Datetime

from datetime import datetime, timedelta

now = datetime.now()
print(now)

yesterday = now - timedelta(days=1)

print(yesterday)
```

```python
## time 

import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## Regular expression

import re

pattern = '\d+'
text = 'There are 123 apples 456'
match = re.search(pattern,text)
print(match.group())
```
