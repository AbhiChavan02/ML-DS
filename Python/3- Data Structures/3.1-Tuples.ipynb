```python
## Tuple Methods

numbers

#1. Count - How many no. of times number is present
print(numbers.count(1))

#2. Index - on which number the element is present
print(numbers.index(3))
```

```python
## Packing and Unpacking Tuples

# Packing - When we define comma separated by default it going to pack them in tuple itself
packed_tuple = 1, "hello", 3.14
print(packed_tuple)
```

```python
# Unpacking - We gonna create variable for how many element in tuple 
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# Unpacking with * - It take all the element till last element
numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
## Nested Tuple - Tuple inside a tuple is called nested tuple
tpl=[(1,2,3,4),(6,7,8,9),(1,"Hello",3.14,"c")]
tpl[2][0:3]

## Nested List - 
# lst=[[1,2,3,4],[6,7,8,9],[1,"Hello",3.14,"c"]]
# lst[0][0:3]
```

```python
nested_tuple = ((1,2,3),("A","B","C"),(True,False))

## Access the Elements inside a Tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over a nested tuple
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
