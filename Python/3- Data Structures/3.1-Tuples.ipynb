```python
## Packing and Unpacking Tuples

# Packing - When we define comma separated by default it going to pack them in tuple itself
packed_tuple = 1, "hello", 3.14
print(packed_tuple)
```

```python
# Unpacking - We gonna create variable for how many element in tuple 
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# Unpacking with * - It take all the element till last element
numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
nested_tuple = ((1,2,3),("A","B","C"),(True,False))

## Access the Elements inside a Tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over a nested tuple
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```

15. Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.


```python
def min_tuple(tpl):
    return min(tpl)
def max_tuple(tpl):
    return max(tpl)
def sum_tuple(tpl):
    return sum(tpl)

sample_tpl = (1,2,3,4,5)
print(f"Minimum: {min_tuple(sample_tpl)}")
print(f"Maximun: {max_tuple(sample_tpl)}")
print(f"Sum of all Tuple: {sum_tuple(sample_tpl)}")
```
