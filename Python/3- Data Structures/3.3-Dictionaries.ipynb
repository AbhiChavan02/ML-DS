#### Introduction to Dictionaries


Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (eg. strings, numbers, or tuples), while values can be of any type.


```python
## Creating Dictionaries

# Two ways of creating dictionaries

empty_dictionary={}
print(type(empty_dictionary))
```

```python
empty_dict = dict()
empty_dict
```

```python
student={"name":"Abhi","age":23,"grade":88}
print(student)
print(type(student))
```

```python
# Single key is always used
student={"name":"Abhi","age":23,"name":88}
print(student)
```

same like mysql data will be stored in key value pair


```python
## Accessing Dictionary Elements

student={"name":"Abhi","age":23,"grade":'A'}
print(student)
```

```python
## Accessing Dictionary Elements - Easiest
print(student['grade'])
print(student['age'])

## Accessing Dictionary element by get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name', "Not Available"))
```

print(student.get('last_name', "Not Available"))

"Not Available" is default value


```python
## Modifying Dictionary Elements
# Dictionary are mutable, so we can add, update or delete elements but all times our key need to be unique.

print(student)
```

```python
# 1. Update - Change or update
student["age"] = 25
print(student)
# 2. Add - New key and value
student["address"] = "Pune"
print(student)
```

```python
# 3. Delete - This will specifically delete key and value pair
del student['grade']
print(student)
```

```python
## Dictionary Methods

# 1. Keys() methods - If I want to get all keys of data
keys = student.keys()
print(keys)

# 2. Values() methods - If I want to get all values of data
values = student.values()
print(values)

# 3. Items() methods - If I want all key value pairs
items = student.items()
print(items)
```

```python
## Shallow copy - VVIP points

student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Abhijeet"
print(student)
print(student_copy)
```

##### Shallow Copy -

This is basically allocated memory. it will take this variable and it will give the reference of an variable to other memory which will having that entire value that we really want to store.


```python
student_copy1 = student.copy()
print(student_copy1)
print(student)
```

```python
student["name"] = "Aarti"
print(student_copy1)
print(student)
```

```python
### Iterating Over Dictionaries
## You can use loops to iterate over dictionaries, keys, values, items(key:value) pair

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for values in student.values():
    print(values)
```

```python
## Iterate over (Items) key value pair
for keys, values in student.items():
    print(f"{keys}:{values}")
```

```python
## Nested Dictionaries - dictionary inside another dictionary
students = {
    "Student1":{"name":"Abhi", "age":23},
    "Student2":{"name":"Aarti", "age":21}
}

print(students)
```

```python
## Access Nested Dictionary elements
print(students["Student2"]["name"])
print(students["Student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over Nested Dictionaries
for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comphrehensions

squares = { x:x**2 for x in range(5)}
print(squares)
```

x : key

x\*\*2 : value


```python
## Conditional dictionary comprehension

evens = { x:x**2 for x in range(10) if x%2==0 }
print(evens)
```

```python
## Practical Examples

## 1. Use a dictionary to count the frequency of elements in list

numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## 2. Merge 2 Dictionaries into one
dict1 = {"a":1, "b":2}
dict2 = {"b":3, "c":4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
