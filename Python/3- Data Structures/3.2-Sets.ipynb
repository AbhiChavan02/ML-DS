#### Mathematical Operations


```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

### Union
union_set = set1.union(set2)
print(union_set)

### Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)

### Intersection Update
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

### Difference
print(set1.difference(set2))
```

```python
## Symmetric Difference - from set1 and set2 which are the common elements only that will be remote and remaining all the unique element both set will be combined 

set1.symmetric_difference(set2)
```

```python
## Sets Methods
set1 = {1,2,3,4,5}
set2 = {3,4,5}

#1. is subset - in parent set which has the all emements not available in set1 so it defines in bool
print(set1.issubset(set2))

#2. is superset - whether it has all the element of set2
print(set1.issuperset(set2))
```

3. Create two sets: one with the first 5 positive integers and another with the first 5 even integers. Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.


```python
set1 = set(range(1,6))
set2 = set(range(2,11,2))
print(f"Union: {set1 | set2}")
print(f"Intersection: {set1 & set2}")
print(f"Difference (set1 - set2): {set1 - set2}")
print(f"Symmetric Difference: {set1 ^ set2}")
```
