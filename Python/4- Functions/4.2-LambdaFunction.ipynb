##### Lambda Functions


Lambda Functions are small anonymous functions defined using the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.


```python
# Syntax
lambda argument: expression
```

```python
def addition(a,b):
    return a+b

addition(2,7)
```

If we have one expression and we have multiple parameters then instead of wholesome function here we create anonymous function which is also called as lambda function


```python
addition = lambda a,b:a+b
type(addition)
print(addition(5,6))
```

```python
def even(num):
    if num%2==0:
        return True

even(24)
```

```python
even1 = lambda num:num%2==0
even1(12)
```

```python
def add(x,y,z):
    return x+y+z

add(6,7,9)
```

```python
add1 = lambda x,y,z:x+y+z

add1(12,15,12)
```

```python
# Map() - applies a function to all items in a list
num = [1,2,3,4,5,6]
def square(numbers):
    return numbers**2

square(4)
```

```python
list(map(lambda x:x**2, num))
```
