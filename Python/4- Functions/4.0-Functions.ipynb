##### Introduction to Functions


A function is block of code that performs a specific task. Functions help in organizing code, resuing code, and imporving readiablity.


```python
## Syntax
def function_name(parameters):
    """Docstring"""
    # Function body
    return expression
```

```python
## Why Function?

num = 24
if num%2==0:
    print("This number is even")
else:
    print("This number is odd")
```

If we need to reuse the piece of code to multiple places then we use functions


```python
## Creating a function
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("This number is even")
    else:
        print("This number is odd")
```

```python
## Call above function
even_or_odd(24)
```

```python
## Function with multiple parameters

def add(a,b):
    return a + b

result=add(2,4)

print(result)
```

```python
## Default Paramters - this function is responsible for parameters. so if we not giving default value to the name so we

def greet(name):
    print(f"Hello {name} welcome to paradise")

greet("Abhi")
```

```python
## This is how we assign default paramters
def greet(name="Guest"):
    print(f"Hello {name} welcome to paradise")

greet("Abhi")
```

```python
## Variable length arguments
## Positional and Keywords arguments

# We have in a function many many parameter as such so we use "*args"

## Postional Arguments - It is a set of parameters and only the one key so are called as positional arguments
def print_num(*args):
    for num in args:
        print(num)


print_num(1,2,3,4,5,"Abhi")
```

```python
## Keywords Arguments - is given by "**kwargs" in it all the keys or all the parameters are in key-value pair

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="Abhi",age="32", country="India")
```

```python
## Combining both keywords and positional arguments but in it postional arguments cannot appear after keyword and vice-versa

def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional Arguments: {val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(1,2,3,4,5,"Abhi",name="Abhi",age="32", country="India")
```

```python
## Return Statements - In it our function returns something after the entire statement is executed

def multiply(a,b):
    return a * b

multiply(2,3)
```

```python
## Return multiple parameters from functions
def multiply(a,b):
    return a * b, a

multiply(2,3)
```
